```python
def read_in_taf_rates(years):
    """Read in and concatenate annual TAF rates datasets."""
    dfs = []
    for year in years:
        df = pd.read_csv(f'/gpfs/data/cms-share/duas/56930/Nadia/data/medicaid_NF_pmt_rates/calculated_rates/per_diems/taf_allowed_{year}.csv')
        df['filetype'] = 'taf'
        df['year'] = year
        df['method'] = 'allowed_amount'
        dfs.append(df)
    return pd.concat(dfs, ignore_index=True)
    
test_taf = read_in_taf_rates([2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021])

```

```python
def read_in_max_rates(years):
    """Read in and concatenate annual MAX rates datasets."""
    dfs = []
    for year in years:
        df = pd.read_csv(f'/gpfs/data/cms-share/duas/56930/Nadia/data/medicaid_NF_pmt_rates/calculated_rates/per_diems/max_estimated_{year}_2026-04-28.csv')
        dfs.append(df)
        if year == 2013: 
            df = df.loc[~df['state'].isin(['MA','NY'])]
        elif year == 2014: 
            df = df.loc[~df['state'].isin(['AK','AL','CO','DE','KS','MD','MT','NM','NY','RI','WI'])]
    return pd.concat(dfs, ignore_index=True)

test_max = read_in_max_rates([2011, 2012, 2013, 2014, 2015])
print(test_max['state'].value_counts())
```

```python
max_taf = pd.concat([test_max, test_taf], axis=0)
```
